# Gradient viewer
### Compute Sobel x and y gradients, gradient magnitude and direction, and the Laplacian of one image

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
gray=cv2.imread('pic.jpg',0)
```

```python
gx=cv2.Sobel(gray,cv2.CV_64F,1,0,ksize=3)
gy=cv2.Sobel(gray,cv2.CV_64F,0,1,ksize=3)

magnitude,angle=cv2.cartToPolar(gx, gy, angleInDegrees=True)

lap=cv2.Laplacian(gray,cv2.CV_64F)


```

```python
gx_show=cv2.convertScaleAbs(gx)
gy_show=cv2.convertScaleAbs(gy)
mag_show=cv2.convertScaleAbs(magnitude)
lap_show=cv2.convertScaleAbs(lap)
```

```python
fig,axes=plt.subplots(1,5,figsize=(20,5))

axes[0].imshow(gx_show,cmap='gray');axes[0].set_title('sobel_x')
axes[1].imshow(gy_show,cmap='gray');axes[1].set_title('sobel_y')
axes[2].imshow(mag_show,cmap='gray');axes[2].set_title('magnitude')
axes[3].imshow(angle,cmap='hsv');axes[3].set_title('direction')
axes[4].imshow(lap_show,cmap='gray');axes[4].set_title('laplacian')

for ax in axes:
    ax.axis('off')

plt.tight_layout()
plt.savefig('p2_figure.jpg')
plt.show()
```
