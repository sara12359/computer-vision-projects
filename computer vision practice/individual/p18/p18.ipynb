# Robust image loader
### Load an image from a path given on the command line and show it in a window titled 'Image Display'. Wait for a keypress, then close. Handle a missing file, an unreadable file and a non-image file.

```python
import cv2
import os
import sys
```

```python
path='pic.jpg'

#if the image is missing
if not os.path.exists(path):
    print('Erorr: the image is not found')
    sys.exit(1)

#check if its not a file
if not os.path.isfile(path):
    print('Erorr: cannot read this its not a file')
    sys.exit(1)

img=cv2.imread(path)

#check if its an image
if img is None:
    print('Erorr : it not an image')
    sys.exit(1)

cv2.imshow('Image Display',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
