# Video player and recorder
### Play a video file, then the webcam, with cv2.VideoCapture(). Overlay a frame counter and FPS. Press s to start/stop recording a clip and q to quit.


```python
import cv2
import time
```

```python
cap=cv2.VideoCapture('video.mp4')

while True:
    ok,frame=cap.read()
    if not ok:
        break
    cv2.imshow('video',frame)
    if cv2.waitKey(30)== ord('q'):
        break
cap.release()
cv2.destroyAllWindows()
```
