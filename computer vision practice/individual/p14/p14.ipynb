# Image viewer and saver
### Load an image with cv2.imread(), display it, resize it to half and to double size, and save the original as JPG, PNG and BMP.

```python
import cv2
import os
import matplotlib.pyplot as plt
```

```python
img=cv2.imread('pic.jpg')
print('shape:',img.shape)
print('data type:',img.dtype)
```

```python
half=cv2.resize(img,None,fx=0.5,fy=0.5,interpolation=cv2.INTER_AREA)
double=cv2.resize(img,None,fx=2,fy=2,interpolation=cv2.INTER_CUBIC)

print('original size:',img.shape)
print('half size:',half.shape)
print('double size:',double.shape)
```

```python
cv2.imwrite('half_size.jpg',half)
cv2.imwrite('double_size.jpg',double)
```

```python
cv2.imwrite('img.jpg',img)
cv2.imwrite('img.png',img)
cv2.imwrite('img.bmp',img)
```

```python
files=['img.jpg','img.png','img.bmp']
for f in files:
    check=cv2.imread(f)
    kb=os.path.getsize(f)/1024
    print(f,'|Shape:',check.shape,'|data type',check.dtype,'| size:',kb)
```

```python
fig,axes=plt.subplots(1,3,figsize=(15,5))
axes[0].imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
axes[0].set_title('Original '+str(img.shape[:2]))

axes[1].imshow(cv2.cvtColor(half,cv2.COLOR_BGR2RGB))
axes[1].set_title('Half'+str(half.shape[:2]))

axes[2].imshow(cv2.cvtColor(double,cv2.COLOR_BGR2RGB))
axes[2].set_title('Double'+str(double.shape[:2]))

plt.tight_layout()
plt.savefig('p14-figure.jpg')
plt.show()
```
