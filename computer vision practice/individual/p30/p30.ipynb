# Pixel art flag generator
### Draw a national flag (e.g. Rwanda) on a blank canvas using only cv2.rectangle, cv2.circle, cv2.line, cv2.fillPoly and cv2.putText.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
BLUE   = (222, 161, 0)
YELLOW = (1, 209, 250)
GREEN  = (61, 96, 32)
GOLD   = (0, 215, 255)
WHITE  = (255, 255, 255)
BLACK  = (0, 0, 0)
```

```python
#creating the white falg 
flag=np.full((450,600,3),255,dtype=np.uint8)
plt.imshow(cv2.cvtColor(flag,cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
#adding the color blue
cv2.rectangle(flag,(0,0),(599,199),BLUE,-1)
plt.imshow(cv2.cvtColor(flag,cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
#adding the yellow
cv2.rectangle(flag,(0,200),(599,324),YELLOW,-1)
plt.imshow(cv2.cvtColor(flag,cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
cv2.rectangle(flag,(0,325),(599,449),GREEN,-1)
plt.imshow(cv2.cvtColor(flag,cv2.COLOR_BGR2RGB))    
plt.axis('off')
plt.show()
```

```python
# adding the sun circle
cx,cy=480,100
cv2.circle(flag,(cx,cy),25,GOLD,-1)
plt.imshow(cv2.cvtColor(flag,cv2.COLOR_BGR2RGB))    
plt.axis('off')
plt.show()
```

```python
for i in range(24):
    angle = i * 2 * np.pi / 24
    x1 = int(cx + 35 * np.cos(angle))
    y1 = int(cy + 35 * np.sin(angle))
    x2 = int(cx + 60 * np.cos(angle))
    y2 = int(cy + 60 * np.sin(angle))
    cv2.line(flag, (x1, y1), (x2, y2), GOLD, 3)

plt.imshow(cv2.cvtColor(flag,cv2.COLOR_BGR2RGB))    
plt.axis('off')
plt.show()
```

```python
# caption below the flag
cv2.putText(flag, "Rwanda", (240, 435), cv2.FONT_HERSHEY_SIMPLEX, 1, BLACK, 2)

cv2.imwrite('RwandaFlag.jpg',flag)

plt.imshow(cv2.cvtColor(flag,cv2.COLOR_BGR2RGB))    
plt.axis('off')
plt.show()
```
