```python
import numpy as np
import cv2
import matplotlib.pyplot  as plt
```

### gray gradient from black to white

```python
row=np.linspace(0,255,200)
gradient=np.tile(row,(200,1)).astype(np.uint8)
cv2.imshow('Gradient', gradient)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### chessboard

```python

small = np.indices((8, 8)).sum(axis=0) % 2

big=np.kron(small,np.ones((25,25)))

chessboard = (big * 255).astype(np.uint8)

cv2.imshow('chessboard',chessboard)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

### Red to Blue gradient

```python
color=np.zeros((200,200,3),dtype=np.uint8)

row=np.linspace(0,255,200)

color[:,:,0]=row[::-1]
color[:,:,1]=0
color[:,:,2]=row

cv2.imshow('red to blue',color)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(12, 4))

axes[0].imshow(gradient, cmap='gray', vmin=0, vmax=255)
axes[0].set_title('Grey Gradient')

axes[1].imshow(chessboard, cmap='gray', vmin=0, vmax=255)
axes[1].set_title('Chessboard 8x8')

axes[2].imshow(color)
axes[2].set_title('Red to Blue Gradient')

for ax in axes:
    ax.axis('off')

plt.tight_layout()
plt.savefig('p4_figure.png')
plt.show()
```

```python
cv2.imwrite('GrayGradient.jpg',gradient)

cv2.imwrite("Chessboard.jpg",chessboard)

cv2.imwrite('WrongRedToBlue.jpg',color)
cv2.imwrite('RedToBlue.jpg',cv2.cvtColor(color,cv2.COLOR_RGB2BGR))
```
