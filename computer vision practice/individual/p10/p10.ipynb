### Convert one photo to RGB, grey, HSV and LAB, and split every colour space into its channels.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python

img=cv2.imread('pic.jpg')
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title('original image')
plt.axis('off')
plt.show()
```

```python
rgb=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
hsv=cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
lab=cv2.cvtColor(img,cv2.COLOR_BGR2LAB)
```

```python
r,g,b=cv2.split(rgb)
h,s,v=cv2.split(hsv)
l,a,b2=cv2.split(lab)
```

```python
for name, x in [('R',r),('G',g),('B',b),('H',h),('S',s),('V',v),('L',l),('a',a),('b',b2)]:
    print(name ,'min',x.min(),'max',x.max())
```

```python
fig,axes=plt.subplots(4,4,figsize=(14,14))

#RGB
axes[0,0].imshow(rgb)
axes[0,0].set_title('RGB')

axes[0,1].imshow(r,cmap='gray')
axes[0,1].set_title('R')

axes[0,2].imshow(g,cmap='gray')
axes[0,2].set_title('G')

axes[0,3].imshow(b,cmap='gray')
axes[0,3].set_title('B')

#grayscale
axes[1,0].imshow(gray,cmap='gray')
axes[1,0].set_title('GrayScale')

#HSV
axes[2,0].imshow(hsv)
axes[2,0].set_title('HSV')

axes[2,1].imshow(h,cmap='gray')
axes[2,1].set_title('Hue')

axes[2,2].imshow(s,cmap='gray')
axes[2,2].set_title('Saturation')

axes[2,3].imshow(v,cmap='gray')
axes[2,3].set_title('Value')

#lab
axes[3,0].imshow(lab)
axes[3,0].set_title('LAB')

axes[3,1].imshow(l,cmap='gray')
axes[3,1].set_title('Lightness')

axes[3,2].imshow(a,cmap='gray')
axes[3,2].set_title('a channel')

axes[3,3].imshow(b2,cmap='gray')
axes[3,3].set_title('b channel')

#remove unused pannels
axes[1,1].axis('off')
axes[1,2].axis('off')
axes[1,3].axis('off')

for ax in axes.flatten():
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()
 
# plt.savefig('color_spaces.png')
plt.show()










```
