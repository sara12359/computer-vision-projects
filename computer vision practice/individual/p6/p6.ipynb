```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import textwrap
```

```python
img=cv2.imread('pic.jpg')
cv2.imshow('orginal image',img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
b,g,r=cv2.split(img)

zeros=np.zeros_like(b)
```

```python
swapped=cv2.merge([r,g,b])
no_red=cv2.merge([b,g,zeros])
no_blue=cv2.merge([zeros,g,r])
no_green=cv2.merge([b,zeros,r])

big_g=g.astype(np.float32)*1.5
big_g=np.clip(big_g,0,255).astype(np.uint8)
more_green=cv2.merge([b,big_g,r])
```

```python
swapped_rgb=cv2.cvtColor(swapped,cv2.COLOR_BGR2RGB)
no_red_rgb=cv2.cvtColor(no_red,cv2.COLOR_BGR2RGB)
no_blue_rgb=cv2.cvtColor(no_blue,cv2.COLOR_BGR2RGB)
no_green_rgb=cv2.cvtColor(no_green,cv2.COLOR_BGR2RGB)
more_green_rgb=cv2.cvtColor(more_green,cv2.COLOR_BGR2RGB)
original=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
```

```python
fig,axes=plt.subplots(2,3,figsize=(12,8))

axes[0,0].imshow(original)
axes[0,0].set_title('original image')

axes[0,1].imshow(swapped_rgb)
axes[0,1].set_title('R Swapped with B image')


axes[0,2].imshow(more_green_rgb)
axes[0,2].set_title('50 precante more green image')


axes[1,0].imshow(no_red_rgb)
axes[1,0].set_title('NO Red image')


axes[1,1].imshow(no_blue_rgb)
axes[1,1].set_title('NO Blue image')

axes[1,2].imshow(no_green_rgb)
axes[1,2].set_title('no green image')

for ax in axes.flatten():
    ax.set_xticks([])
    ax.set_yticks([])

plt.tight_layout()

plt.savefig('p6_channel_mixer.jpg')
plt.show()

```
